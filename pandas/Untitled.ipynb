```python
top_channels = df['channel_name'].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_channels.values,
    y=top_channels.index,
    palette='viridis'
)

plt.title("Top 10 Trending Channels")
plt.xlabel("Number of Videos")
plt.ylabel("Channel Name")

plt.show()
```
